### **K-Means Clustering**

An algorithm used for **unlabeled data** that groups points based on similarity. The **Elbow Method** is used to choose the optimal $K$ value (number of clusters).

**Euclidean Distance Formula:**


$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

**Algorithm Steps:**

1. **Choose $K$:** Decide the number of clusters (e.g., $K = 2$).
2. **Initialize Centroids:** Randomly assign initial cluster centroids.
3. **Calculate Distance:** Compute the Euclidean distance between each data point and the centroids.
4. **Assign Clusters:** Group each data point with the nearest centroid (minimum distance).
5. **Update Centroids:** Calculate the average of each cluster to assign new centroids.
6. **Repeat:** Continue steps 3–5 until the centroids no longer change.

### Setup

* **$K = 2$** (Two clusters)
* **Initial Centroids (highlighted in red):**
* $C_1 = (1, 1)$ (Subject 1)
* $C_2 = (5, 7)$ (Subject 4)


* **Euclidean Distance Formula:** $d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$

**Subject 3 is equidistant ($3.61$) from both centroids; it is assigned to Cluster 1 here.*

**Update Centroids (Take the average of each cluster):**

* **New $C_1$** (Subjects 1, 2, 3):

$$C_2 = \left(\frac{5 + 3.5 + 4.5 + 3.5}{4}, \frac{7 + 5 + 5 + 4.5}{4}\right) = (4.13, 5.38)$$

# Step 14 - Standardize the input features

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_features = scaler.fit_transform(df)
```

# Step 15 - Use elbow method

```python
from sklearn.cluster import KMeans

inertia_values = []
for i in range(1, 11):
    model = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
        )
    model.fit(scaled_features)
    inertia_values.append(model.inertia_)
```

# Step 16 - Plot elbow graph

```python
from matplotlib.pyplot import figure

figure(figsize=(7, 4))

plt.plot(
    range(1, 11),
    inertia_values,
    marker="o"
  )

plt.xlabel("Number of clusters")
plt.ylabel("Inertia")
plt.title("Elbow method")
plt.show()
```

# step 17 - Train K-Means with 3 Clusers

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
    )
iris_clusters = kmeans.fit_predict(scaled_features)
```

# Step 18 - Add cluster labels to the dataset

```python
df_clusters = df.copy()
df_clusters["cluster"] = kmeans.labels_
df_clusters.head()
```

# Step 19 - Plot Cluster

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df_clusters,
    x="petal_length",
    y="petal_width",
    hue="cluster",
    palette="viridis",
    s=100
)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("K-Means Clustering on Iris Data")
plt.tight_layout()
plt.show()
```

## Evaluation Methods

# Step 20 - Calculate Sillhouette score

```python
from sklearn.metrics import silhouette_score

score= silhouette_score(
    scaled_features,
    iris_clusters
    )
print("Silhouette Score:",score)
```
