# **Handling Outliers**

* **Minimum**: Smallest value in the dataset excluding outliers.


* **First Quartile ($Q_1$)**: Middle value of the lower half of the data (25th percentile).


* **Median ($Q_2$)**: Middle value that splits the dataset into two equal parts (50th percentile).


* **Third Quartile ($Q_3$)**: Middle value of the upper half of the data (75th percentile).


* **Maximum**: Largest value in the dataset excluding outliers.


* **IQR Formula**: $IQR = Q_3 - Q_1$ (the range between $Q_1$ and $Q_3$).


* **Lower Fence**: $Q_1 - 1.5 \times IQR$.


* **Upper Fence**: $Q_3 + 1.5 \times IQR$.


* **IQR Outlier Rule**: Any data point below the lower fence or above the upper fence is classified as an outlier.

# **PassengerId Outlier Calculation**

* **Given Values:**
* **Minimum:** $1.0$


* **$Q_1$ (25%):** $223.5$


* **Median ($Q_2$, 50%):** $446.0$


* **$Q_3$ (75%):** $668.5$


* **Maximum:** $891.0$




* **Step-by-Step Formulas:**
* **IQR:** $Q_3 - Q_1 = 668.5 - 223.5 = \mathbf{445.0}$


* **Lower Fence:** $Q_1 - 1.5 \times IQR = 223.5 - (1.5 \times 445.0) = 223.5 - 667.5 = \mathbf{-444.0}$


* **Upper Fence:** $Q_3 + 1.5 \times IQR = 668.5 + (1.5 \times 445.0) = 668.5 + 667.5 = \mathbf{1336.0}$




* **Result:** **No Outliers.** Since the minimum ($1.0$) is above $-444.0$ and the maximum ($891.0$) is below $1336.0$, all values fall within the fences.

# **Visualization**

* **Definition**: Presenting data in graphical or visual form to make it clearer and easier to understand.


* **Key Benefits**:
* Makes it easy to analyze and summarize data.


* Helps easily understand data features.


* Aids in identifying trends or patterns.


* Helps extract meaningful insights from the data.

# **Types of Data Visualizations**

* **Histogram**: Represents the frequency distribution of the data.


* **Bar Chart**: Represents the total observations in the data for a particular category.


* **Box Plot**: Displays data distribution based on the five-number summary (minimum, first quartile, median, third quartile, maximum).


* **Pie Chart**: Represents the percentage of data for each category.


* **Pair-plot**: Shows the bivariate distribution of datasets and pairwise relationships between variables.


* **Scatter Plot**: Represents the relationship and correlation between two numerical variables.


* **Violin Plot**: Used to plot numeric data.


* **Line Chart**: Tracks changes over periods of time and is used in time series data.


* **Strip Plot**: Draws a scatter plot on a categorical basis.


* **Heatmaps**: Uses color to represent data values in the same way a bar graph uses height and width.
